# Colab Notebook: Download LeafNet75 Masks Dataset
This standalone notebook downloads the LeafNet75 segmentation mask datasets in Google Colab's temporary storage, zips the resulting dataset folder, and downloads `leafnet_masks.zip` directly to your local computer.

```python
# Step 1: Install required packages
!pip install -q datasets huggingface_hub pyarrow pillow
```

```python
# Step 2: Download LeafNet75 lesion and wild masks from Hugging Face
from datasets import load_dataset
from pathlib import Path

target_dir = Path('/content/data/raw/leafnet_masks')
target_dir.mkdir(parents=True, exist_ok=True)

print('Downloading LeafNet75 lesion masks...')
ds_lesion = load_dataset('LeafNet75/Leaf-Disease-Segmentation-masks')
ds_lesion.save_to_disk(str(target_dir / 'lesion_masks'))

print('Downloading LeafNet75 in-the-wild masks...')
ds_wild = load_dataset('LeafNet75/In_the_Wild_masks600')
ds_wild.save_to_disk(str(target_dir / 'wild_masks'))
print(f'Successfully downloaded LeafNet75 mask datasets to {target_dir}')
```

```python
# Step 3: Zip dataset & download directly to your computer
from google.colab import files
import shutil

print('Compressing leafnet_masks.zip...')
shutil.make_archive('/content/leafnet_masks', 'zip', str(target_dir))
print('Triggering browser download for leafnet_masks.zip...')
files.download('/content/leafnet_masks.zip')
```
