# Colab Notebook: Download PlantSeg Dataset
This standalone notebook downloads the PlantSeg dataset (`Voxel51/PlantSeg-Test`) in Google Colab's temporary storage, zips the resulting dataset folder, and downloads `plantseg.zip` directly to your local computer.

```python
# Step 2: Download PlantSeg dataset from Hugging Face
from datasets import load_dataset
from pathlib import Path

target_dir = Path('/content/data/raw/plantseg')
target_dir.mkdir(parents=True, exist_ok=True)

print('Downloading PlantSeg dataset (Voxel51/PlantSeg-Test)...')
ds_plantseg = load_dataset('Voxel51/PlantSeg-Test')
ds_plantseg.save_to_disk(str(target_dir))
print(f'Successfully downloaded {len(ds_plantseg)} samples to {target_dir}')
```

```python
# Step 3: Zip dataset & download directly to your computer
from google.colab import files
import shutil

print('Compressing plantseg.zip...')
shutil.make_archive('/content/plantseg', 'zip', str(target_dir))
print('Triggering browser download for plantseg.zip...')
files.download('/content/plantseg.zip')
```
