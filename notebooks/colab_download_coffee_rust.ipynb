# Colab Notebook: Download Coffee Leaf Rust Dataset
This standalone notebook downloads the Coffee Leaf Rust repository files (`MaryPazRB/Paper_CLR_CV`) from Hugging Face Hub using `snapshot_download`, packages it into `coffee_rust.zip`, and downloads it directly to your local computer.

```python
# Step 1: Install required packages
!pip install -q huggingface_hub datasets pyarrow pillow
```

```python
# Step 2: Download Coffee Leaf Rust repository using snapshot_download
from huggingface_hub import snapshot_download
from pathlib import Path

target_dir = Path('/content/data/raw/coffee_rust')
target_dir.mkdir(parents=True, exist_ok=True)

print('Downloading Coffee Leaf Rust repository (MaryPazRB/Paper_CLR_CV)...')
try:
    snapshot_download(
        repo_id='MaryPazRB/Paper_CLR_CV',
        repo_type='model',
        local_dir=str(target_dir)
    )
    print(f'Successfully downloaded Coffee Rust repository to {target_dir}')
except Exception as e:
    print(f'Model snapshot download error: {e}. Trying space snapshot...')
    snapshot_download(
        repo_id='MaryPazRB/Paper_CLR_CV',
        repo_type='space',
        local_dir=str(target_dir)
    )
```

```python
# Step 3: Zip dataset & download directly to your computer
from google.colab import files
import shutil

print('Compressing coffee_rust.zip...')
shutil.make_archive('/content/coffee_rust', 'zip', str(target_dir))
print('Triggering browser download for coffee_rust.zip...')
files.download('/content/coffee_rust.zip')
```
