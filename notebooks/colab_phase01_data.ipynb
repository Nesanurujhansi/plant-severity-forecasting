# Phase 01: Standalone Colab Dataset Downloaders & Zip Exporter
This standalone notebook allows downloading each dataset individually in Google Colab's temporary storage and downloading the resulting dataset `.zip` file directly to your local computer without using Google Drive storage or uploading your local project folder.

```python
# Cell 1: Install required packages & setup local temp directories
!pip install -q datasets huggingface_hub pyarrow pillow pyyaml pydantic pandas tqdm requests

import os
from pathlib import Path

DATA_DIR = Path('/content/data')
RAW_DIR = DATA_DIR / 'raw'
INTERIM_DIR = DATA_DIR / 'interim'

RAW_DIR.mkdir(parents=True, exist_ok=True)
INTERIM_DIR.mkdir(parents=True, exist_ok=True)
print('Initialized temporary data directory structure at /content/data')
```

```python
# Cell 2: Download PlantSeg Dataset & Export Zip
from datasets import load_dataset
from google.colab import files
import shutil

plantseg_dir = RAW_DIR / 'plantseg'
print('Downloading PlantSeg test subset (Voxel51/PlantSeg-Test)...')
ds_plantseg = load_dataset('Voxel51/PlantSeg-Test')
ds_plantseg.save_to_disk(str(plantseg_dir))
print(f'Downloaded {len(ds_plantseg)} samples to {plantseg_dir}')

print('Compressing plantseg.zip...')
shutil.make_archive('/content/plantseg', 'zip', str(plantseg_dir))
print('Triggering browser download for plantseg.zip...')
files.download('/content/plantseg.zip')
```

```python
# Cell 3: Download Coffee Leaf Rust Dataset & Export Zip
coffee_dir = RAW_DIR / 'coffee_rust'
print('Downloading Coffee Leaf Rust dataset (MaryPazRB/Paper_CLR_CV)...')
ds_coffee = load_dataset('MaryPazRB/Paper_CLR_CV')
ds_coffee.save_to_disk(str(coffee_dir))
print(f'Downloaded Coffee Rust dataset to {coffee_dir}')

print('Compressing coffee_rust.zip...')
shutil.make_archive('/content/coffee_rust', 'zip', str(coffee_dir))
print('Triggering browser download for coffee_rust.zip...')
files.download('/content/coffee_rust.zip')
```

```python
# Cell 4: Download LeafNet75 Segmentation Masks & Export Zip
leafnet_masks_dir = RAW_DIR / 'leafnet_masks'
print('Downloading LeafNet75 lesion and wild masks...')
ds_lesion = load_dataset('LeafNet75/Leaf-Disease-Segmentation-masks')
ds_lesion.save_to_disk(str(leafnet_masks_dir / 'lesion_masks'))
ds_wild = load_dataset('LeafNet75/In_the_Wild_masks600')
ds_wild.save_to_disk(str(leafnet_masks_dir / 'wild_masks'))
print(f'Downloaded LeafNet75 masks to {leafnet_masks_dir}')

print('Compressing leafnet_masks.zip...')
shutil.make_archive('/content/leafnet_masks', 'zip', str(leafnet_masks_dir))
print('Triggering browser download for leafnet_masks.zip...')
files.download('/content/leafnet_masks.zip')
```

```python
# Cell 5: Download PlantVillage Dataset (Optional) & Export Zip
plantvillage_dir = RAW_DIR / 'plantvillage'
print('Downloading PlantVillage dataset (mohanty/PlantVillage)...')
ds_pv = load_dataset('mohanty/PlantVillage')
ds_pv.save_to_disk(str(plantvillage_dir))
print(f'Downloaded PlantVillage dataset to {plantvillage_dir}')

print('Compressing plantvillage.zip...')
shutil.make_archive('/content/plantvillage', 'zip', str(plantvillage_dir))
print('Triggering browser download for plantvillage.zip...')
files.download('/content/plantvillage.zip')
```

```python
# Cell 6: Generate Download Validation Report & Export Zip
import json

summary = {
    'datasets_downloaded': ['plantseg', 'coffee_rust', 'leafnet_masks', 'plantvillage'],
    'status': 'ALL_DOWNLOADED_SUCCESSFULLY'
}

val_json_path = INTERIM_DIR / 'download_validation.json'
val_md_path = INTERIM_DIR / 'download_validation.md'

with open(val_json_path, 'w') as f:
    json.dump(summary, f, indent=2)

val_md_path.write_text('# Colab Ingestion Complete\nAll datasets processed into downloadable zip archives.')

print('Compressing validation_reports.zip...')
shutil.make_archive('/content/validation_reports', 'zip', str(INTERIM_DIR))
files.download('/content/validation_reports.zip')
```
