# Colab Notebook: Download PlantVillage Dataset (Optional)
This standalone notebook downloads the PlantVillage dataset (`mohanty/PlantVillage` with `name='default'`) in Google Colab's temporary storage, zips the resulting dataset folder, and downloads `plantvillage.zip` directly to your local computer.

```python
# Step 2: Download PlantVillage dataset using 'default' config name
from datasets import load_dataset
from pathlib import Path

target_dir = Path('/content/data/raw/plantvillage')
target_dir.mkdir(parents=True, exist_ok=True)

print('Downloading PlantVillage dataset (mohanty/PlantVillage, name="default")...')
try:
    ds_pv = load_dataset('mohanty/PlantVillage', name='default')
    ds_pv.save_to_disk(str(target_dir))
    print(f'Successfully downloaded PlantVillage dataset to {target_dir}')
except Exception as e:
    print(f'Config default error: {e}. Trying fallback...')
    ds_pv = load_dataset('mohanty/PlantVillage')
    ds_pv.save_to_disk(str(target_dir))
    print(f'Successfully downloaded PlantVillage dataset to {target_dir}')
```

```python
# Step 3: Zip dataset & download directly to your computer
from google.colab import files
import shutil

print('Compressing plantvillage.zip...')
shutil.make_archive('/content/plantvillage', 'zip', str(target_dir))
print('Triggering browser download for plantvillage.zip...')
files.download('/content/plantvillage.zip')
```
